```python

from pathlib import Path


# Source code file extensions
SOURCE_EXTENSIONS = {
    ".py",
    ".js",
    ".ts",
    ".tsx",
    ".jsx",
    ".java",
    ".go",
    ".rs",
    ".cs",
    ".php",
}


# Documentation file extensions
DOC_EXTENSIONS = {
    ".md",
    ".txt",
    ".rst",
}


# Config file extensions
CONFIG_EXTENSIONS = {
    ".json",
    ".yaml",
    ".yml",
    ".toml",
    ".ini",
    ".cfg",
    ".env.example",
}


# Common generated / lock files
GENERATED_FILES = {
    "package-lock.json",
    "yarn.lock",
    "pnpm-lock.yaml",
    "poetry.lock",
    "pipfile.lock",
}


# Infrastructure-related folders
INFRA_PREFIXES = (
    ".github/",
    "docker/",
    "infra/",
    "terraform/",
    "k8s/",
)


def classify_file(filename: str) -> str:
    """
    Classify a file into a review category.

    Returns:
        - source
        - test
        - docs
        - config
        - infrastructure
        - generated
        - unknown
    """

    # Convert path to lowercase for safer matching
    path = filename.lower()

    # Extract filename only
    name = Path(path).name

    # Extract extension
    suffix = Path(path).suffix

    # Check generated files first
    if is_generated_file(path):
        return "generated"

    # Infrastructure folders
    if path.startswith(INFRA_PREFIXES):
        return "infrastructure"

    # Test files
    if is_test_file(path):
        return "test"

    # Documentation files
    if suffix in DOC_EXTENSIONS:
        return "docs"

    # Config files
    if suffix in CONFIG_EXTENSIONS:
        return "config"

    # Source code files
    if suffix in SOURCE_EXTENSIONS:
        return "source"

    # Unknown file type
    return "unknown"


def is_test_file(filename: str) -> bool:
    """
    Detect whether a file is a test file.
    """

    name = Path(filename).name

    return (
        filename.startswith("tests/")
        or "/tests/" in filename
        or name.startswith("test_")
        or name.endswith("_test.py")
        or name.endswith(".test.js")
        or name.endswith(".test.ts")
        or name.endswith(".spec.js")
        or name.endswith(".spec.ts")
    )


def is_generated_file(filename: str) -> bool:
    """
    Detect generated files that should usually be ignored.
    """

    name = Path(filename).name

    # Exact generated file names
    if name in GENERATED_FILES:
        return True

    # Minified frontend assets
    if filename.endswith(".min.js") or filename.endswith(".min.css"):
        return True

    # Build output folders
    if "dist/" in filename or "build/" in filename:
        return True

    return False


def should_ignore_file(filename: str) -> bool:
    """
    Determine if a file should be ignored entirely.
    """

    return classify_file(filename) == "generated"

```

```python

from pydantic import BaseModel


class DiffChunk(BaseModel):
    file_path: str
    chunk_index: int
    start_line: int | None = None
    end_line: int | None = None
    content: str


chunk1 = DiffChunk(
    file_path="src/app/example.py",
    chunk_index=0,
    start_line=10,
    end_line=20,
    content="def example():\n    pass\n"
)

chunk2 = DiffChunk(
    file_path="src/app/example1.py",
    chunk_index=1,
    start_line=25,
    end_line=30,
    content="for i in range(5):\n    print(i)\n"
)

chunk3 = DiffChunk(
    file_path="src/app/example2.py",
    chunk_index=0,
    start_line=0,
    end_line=6,
    content="print('This is a new chunk without line numbers')\n"
)

print(chunk1)
print(chunk2)
print(chunk3)
```

```python
class ReviewFileInput(BaseModel):
    filename: str
    file_type: str
    status: str
    additions: int
    deletions: int
    chunks: list[DiffChunk]

review_file_1 = ReviewFileInput(
    filename="src/app/example1.py",
    file_type="source",
    status="modified",
    additions=5,
    deletions=2,
    chunks=[chunk1, chunk2]
)

review_file_2 = ReviewFileInput(
    filename="src/app/example2.py",
    file_type="source",
    status="created",
    additions=6,
    deletions=0,
    chunks=[chunk3]
)


print("review_file_1: ", review_file_1)
print("review_file_2: ", review_file_2)
```

```python
class ReviewInput(BaseModel):
    repo_full_name: str
    pr_number: int
    title: str | None = None
    author: str | None = None
    commit_sha: str | None = None
    files: list[ReviewFileInput]

review_input = ReviewInput(
    repo_full_name="chetan-shah-AI/movie-recommendation-platform",
    pr_number=1,
    title="Initial PR",
    author="chetan-shah-AI",
    commit_sha="sha256=d05f6042130d191a24049f0cd09647e8f289450533d2b3a960c4a32aea0ce67b",
    files=[review_file_1, review_file_2]
)
```

```python
import re

# Sample patch input matching GitHub/Git format
patch_data = """
diff --git a/config.json b/config.json
index 8f23a1c..d4e5f6a 100644
--- a/config.json
+++ b/config.json
@@ -2,6 +2,6 @@
   "appName": "DemoApp",
-  "version": "1.0.0",
+  "version": "1.1.0",
   "environment": "production"
@@ -10,2 +10,3 @@
   "debug": true
+  "port": 8080
"""

# Regular expressions for different patch parts
file_re = re.compile(r"^diff --git a/(.+) b/(.+)")
hunk_re = re.compile(r"^@@ -(\d+),?(\d+)? \+(\d+),?(\d+)? @@")

parsed_changes = []
current_file = None

for line in patch_data.strip().split("\n"):
    # 1. Match file names
    file_match = file_re.match(line)
    if file_match:
        current_file = file_match.group(2) # Target file name
        continue
        
    # 2. Match hunk headers (line numbers)
    hunk_match = hunk_re.match(line)
    if hunk_match:
        old_start, _, new_start, _ = hunk_match.groups()
        continue
        
    # 3. Match added or removed lines
    if line.startswith('+') and not line.startswith('+++'):
        parsed_changes.append({
            "file": current_file,
            "type": "addition",
            "content": line[1:] # Strip the '+' sign
        })
    elif line.startswith('-') and not line.startswith('---'):
        parsed_changes.append({
            "file": current_file,
            "type": "deletion",
            "content": line[1:] # Strip the '-' sign
        })

# Print the structured results
for change in parsed_changes:
    print(f"[{change['type'].upper()}] in {change['file']}: {change['content']}")
```

```python
import re

from app.schemas.review import DiffChunk


HUNK_HEADER_PATTERN = re.compile(
    r"@@ -\d+(?:,\d+)? \+(\d+)(?:,\d+)? @@"
)


def chunk_patch(
    file_path: str,
    patch: str | None,
    max_lines: int = 80,
) -> list[DiffChunk]:
    """
    Split a GitHub patch into smaller chunks.
    """

    if not patch:
        return []

    lines = patch.splitlines()
    chunks: list[DiffChunk] = []

    current_chunk_lines: list[str] = []
    current_start_line: int | None = None
    current_new_line: int | None = None
    chunk_index = 0

    for line in lines:
        hunk_match = HUNK_HEADER_PATTERN.match(line)

        if hunk_match:
            current_new_line = int(hunk_match.group(1))

            if current_start_line is None:
                current_start_line = current_new_line

        if current_new_line is not None and current_start_line is None:
            current_start_line = current_new_line

        current_chunk_lines.append(line)

        if current_new_line is not None:
            if line.startswith("+") and not line.startswith("+++"):
                current_new_line += 1

            elif line.startswith("-") and not line.startswith("---"):
                # Removed lines do not exist in the new file,
                # so the new-file line number does not move.
                pass

            elif not line.startswith("@@"):
                current_new_line += 1

        if len(current_chunk_lines) >= max_lines:
            chunks.append(
                DiffChunk(
                    file_path=file_path,
                    chunk_index=chunk_index,
                    start_line=current_start_line,
                    end_line=current_new_line,
                    content="\n".join(current_chunk_lines),
                )
            )

            chunk_index += 1
            current_chunk_lines = []
            current_start_line = current_new_line

    if current_chunk_lines:
        chunks.append(
            DiffChunk(
                file_path=file_path,
                chunk_index=chunk_index,
                start_line=current_start_line,
                end_line=current_new_line,
                content="\n".join(current_chunk_lines),
            )
        )

    return chunks
```
